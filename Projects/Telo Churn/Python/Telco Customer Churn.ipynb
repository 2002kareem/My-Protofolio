```python
##Import Libraries
```

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
##Read Data
```

```python
file_name = 'Telco Customer Churn.csv'
```

```python
df = pd.read_csv(file_name)
df
```

```python
df.tail()
```

```python
##Show Info About Data
```

```python
df.info()
```

```python
##Clean Data
```

```python
df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan)
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])
```

```python
##Count The Null In Each Columns
```

```python
print(df.isnull().sum())
```

```python
df.dropna(inplace=True)
```

```python
##Count The Null In Each Columns After Remove It . 
```

```python
print(df.isnull().sum())
```

```python
print(f"Number of rows after cleaning: {len(df)}")
```

```python
## Replace Yes And No By 1 And 0 In Churn Column .
```

```python
df['Churn'] = df['Churn'].replace({'Yes': 1, 'No': 0})
```

```python
##The Rate Between Churn or No . 
```

```python
churn_counts = df['Churn'].value_counts(normalize=True) * 100
churn_counts
```

```python
##Figure Of Distribution Of Customer Churn . 
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(x=churn_counts.index, y=churn_counts.values)
plt.title('Distribution of Customer Churn (0: No, 1: Yes)')
plt.ylabel('Percentage')
plt.xticks([0, 1], ['No Churn', 'Churn'])
plt.show()
```

```python
##Some Statical For Some Columns .
```

```python
print("Descriptive Statistics for Numerical Columns")
print(df[['tenure', 'MonthlyCharges', 'TotalCharges']].describe())
```

```python
##Histogram TO Show The Distributon On Numerical Features . 
```

```python
df[['tenure', 'MonthlyCharges', 'TotalCharges']].hist(figsize=(10, 6), bins=30)
plt.suptitle('Distribution of Numerical Features')
plt.tight_layout(rect=[0, 0.03, 1, 0.95]) 
plt.show()
```

```python
##The Defrence Between Contract And Churn . 
```

```python
print("Churn Rate by Contract Type ")
contract_churn = df.groupby('Contract')['Churn'].mean().sort_values(ascending=False)
print(contract_churn)
```

```python
## Figer To Show The Churn Rate By Contract Type . 
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x=contract_churn.index, y=contract_churn.values)
plt.title('Churn Rate by Contract Type')
plt.ylabel('Churn Rate (Proportion of Churn)')
plt.show()
```

```python
## Relationship Between Tenure & Monthly Charges . 
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='tenure', y='MonthlyCharges', hue='Churn', data=df, alpha=0.6)
plt.title('Tenure vs. Monthly Charges, colored by Churn')
plt.xlabel('Tenure (Months)')
plt.ylabel('Monthly Charges')
plt.legend(title='Churn', labels=['No', 'Yes'])
plt.show()
```
